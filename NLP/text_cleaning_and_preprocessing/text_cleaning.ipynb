```python
# Importing the required libraries

import pandas as pd
import re
```

```python
train_data_cleaning_df = pd.read_csv("train_data_cleaning.csv")
```

#### Lowering the case of the text

```python
train_data_cleaning_df["text_lowercase"] = train_data_cleaning_df["text"].apply(lambda x: x.lower())
```

#### Removing the dirty data

Removing the HTML tags

```python
def remove_html_tags(text):
    return re.sub(r"<.*?>", "", text)
```

Removing the URL's

```python
def remove_url(text):
    return re.sub(r"https?://\S+|www\.\S+","",text)
```

Removing Puntuation Marks

```python
def remove_puntuals(text):
    return re.sub(r"[^\w\s]", "", text)
```

```python
train_data_cleaning_df["text_lowercase"] = train_data_cleaning_df["text_lowercase"].apply(remove_puntuals)
```

Chat word Treatment

```python
slang_dict = {"gn": "Good Night", "gm": "Good Morning", "yolo": "You Only Live Once", "asap": "As Soon As Possible"}

slang_data = pd.DataFrame({"slang": ["Hey ! gm", "Chill yaar, yolo", "please do it asap"]})
slang_data
```

```python
def expand_slang(text, slang_dict):
    words = text.split()
    return " ".join([slang_dict.get(word, word) for word in words])
```

Stopwords

```python
from nltk.corpus import stopwords

englist_stop_words = stopwords.words("english")
```

```python
stop_word_text = "Hey, this is my first code and I am coding this to test how well my code is working"
```

```python
def remove_stop_words(text):
    without_stop_words = []
    for i in text.split():
        if i in englist_stop_words:
            without_stop_words.append("")
        else:
            without_stop_words.append(i)
    return " ".join(without_stop_words)
```

```python
remove_stop_words(stop_word_text)
```

```python
train_data_cleaning_df["text_lowercase"] = train_data_cleaning_df["text_lowercase"].apply(remove_stop_words)
```

Removing Emoji's

```python
def remove_emojis(text):
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        u"\U00002600-\U000026FF"  # miscellaneous symbols
        u"\U00002700-\U000027BF"  # dingbats (includes airplane ✈)
        u"\U0001F900-\U0001F9FF"  # supplemental symbols
        u"\U0001FA70-\U0001FAFF"  # extended pictographs
        u"\U0001F3FB-\U0001F3FF"  # skin tone modifiers
        u"\uFE0F"                 # emoji variation selector
    "]+",
    flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
data = {
    'content': [
        "I love pizza 🍕 so much!",
        "Good morning 🌞 🌼",
        "Let's go for a ride 🚗 🏎️",
        "Vacation time 🏖️ ✈️",
        "Happy birthday 🎂 🎉",
        "Feeling strong 💪 🔥"
    ]
}
```

```python
data = pd.DataFrame(data)
data
```

```python
data["content"] = data["content"].apply(remove_emojis)
data
```

```python
for i in data["content"]:
    print(remove_emojis(i))
```

Replacing Emojis

```python
import emoji

def replace_emoji(text):
    return emoji.demojize(text, delimiters=("(", ")"))
```

```python
sample_text = "I love pizza 🍕 so much! Good morning 🌞 🌼 Let's go for a ride 🚗 🏎️ Vacation time 🏖️ ✈️ Happy birthday 🎂 🎉Feeling strong 💪 🔥"
print(replace_emoji(sample_text))
```
