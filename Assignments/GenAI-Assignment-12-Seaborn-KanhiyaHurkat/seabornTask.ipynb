```python
# Importing packages
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Importing Data and Mainupulating

```python
# Readind the CSV data
student_data = pd.read_csv("student_performance_factors_data.csv")

# Changing column type to appropriate one to decrease the memory usage
student_data["Hours_Studied"] = student_data["Hours_Studied"].astype("int8")
student_data["Attendance"] = student_data["Attendance"].astype("int8")
student_data["Parental_Involvement"] = student_data["Parental_Involvement"].astype("category")
student_data["Access_to_Resources"] = student_data["Access_to_Resources"].astype("category")
student_data["Extracurricular_Activities"] = student_data["Extracurricular_Activities"].astype("category")
student_data["Sleep_Hours"] = student_data["Sleep_Hours"].astype("int8")
student_data["Previous_Scores"] = student_data["Previous_Scores"].astype("int8")
student_data["Motivation_Level"] = student_data["Motivation_Level"].astype("category")
student_data["Internet_Access"] = student_data["Internet_Access"].astype("category")
student_data["Tutoring_Sessions"] = student_data["Tutoring_Sessions"].astype("int8")
student_data["Family_Income"] = student_data["Family_Income"].astype("category")
student_data["Teacher_Quality"] = student_data["Teacher_Quality"].astype("category")
student_data["School_Type"] = student_data["School_Type"].astype("category")
student_data["Peer_Influence"] = student_data["Peer_Influence"].astype("category")
student_data["Physical_Activity"] = student_data["Physical_Activity"].astype("int8")
student_data["Learning_Disabilities"] = student_data["Learning_Disabilities"].astype("category")
student_data["Parental_Education_Level"] = student_data["Parental_Education_Level"].astype("category")
student_data["Distance_from_Home"] = student_data["Distance_from_Home"].astype("category")
student_data["Gender"] = student_data["Gender"].astype("category")
student_data["Exam_Score"] = student_data["Exam_Score"].astype("int8")
```

```python
student_data.sample(5)
```

#### Task 1 - Relational Plot

```python
# Relational Plot between 2 numerical column
sns.relplot(student_data, x="Hours_Studied", y="Attendance", kind="line")
plt.show()
```

```python
# Relational Plot between 2 numerical column with gender in hue
sns.relplot(student_data, x="Sleep_Hours", y="Previous_Scores", hue="Gender", kind="line")
plt.show()
```

```python
# Relational Plot with scatter style
sns.relplot(student_data, x="Sleep_Hours", y="Previous_Scores", hue="Gender", kind="scatter")
plt.show()
```

#### Task 2 - Line Plot

```python
# Creating a line plot
sns.lineplot(student_data, x="Hours_Studied", y="Attendance")
plt.show()
```

```python
# In scatter syle
sns.scatterplot(student_data, x="Hours_Studied", y="Attendance", hue="Gender", size="Exam_Score")
plt.show()
```

```python
# Using facet of gender column created the relplot
sns.relplot(student_data, x="Hours_Studied", y="Attendance", col="Gender", kind="scatter")
plt.show()
```

#### Task 3 - Distribution Plots

```python
# Histogram plot
sns.histplot(student_data, x="Previous_Scores")
plt.show()
```

```python
# KDE plot
sns.kdeplot(student_data, x="Previous_Scores", fill=True, hue="Gender")
plt.show()
```

```python
# Rug Plot
sns.rugplot(student_data, x="Previous_Scores")
plt.show()
```

```python
# Combination of Histogram and KDE plot
sns.histplot(student_data, x="Previous_Scores", kde=True)
```

#### Task 4 - Bivariate Distribution  Plots

```python
# Bivariate Histogram plot
sns.histplot(student_data, x="Hours_Studied", y="Attendance")
plt.show()
```

```python
# Bivariate KDE plot
sns.kdeplot(student_data, x="Hours_Studied", y="Attendance")
plt.show()
```

#### Task 5 - Matrix Plots

```python
# Pair plot with gender hue
sns.pairplot(student_data, hue="Gender")
plt.show()
```

```python
# Heatmap by converting data first using pivot_table
dataframe = student_data.pivot_table(index="Parental_Involvement", columns="Sleep_Hours", values="Exam_Score", aggfunc="mean")
sns.heatmap(dataframe, annot=True)
plt.show()
```

#### Task 6 - Categorial Plots

```python
# Bar Plot
sns.barplot(student_data, x="Teacher_Quality", y="Sleep_Hours")
plt.grid()
plt.show()
```

```python
# Box Plot
sns.boxplot(student_data, x="Teacher_Quality", y="Sleep_Hours")
plt.grid()
plt.show()
```

```python
# Violin Plot
sns.violinplot(student_data, x="Teacher_Quality", y="Sleep_Hours")
plt.grid()
plt.show()
```

```python
# Count Plot
sns.countplot(student_data, x="Teacher_Quality")
plt.grid()
plt.show()
```

#### Task 7 - Regression Plots

```python
# Regression plot
sns.regplot(student_data, x="Hours_Studied", y="Exam_Score")
plt.show()
```

```python
# LM Plot 
sns.lmplot(student_data, x="Hours_Studied", y="Exam_Score", hue="Gender")
plt.show()
```
