#### Data Preprocessing

```python
movies_data_df["genres"][0]
```

```python
# Function to exract the name value from the list of genre, cast, etc
# col - Name of the column
# list_size - Pass how much you want to extract the valuen from top,
# if noting passed all values will be added
import ast
def extract_genres(col, list_size = None):
    list_data = []
    data_array = ast.literal_eval(col)
    counter = 0
    list_size = list_size if list_size else len(data_array)
    for i in data_array:
        if counter < list_size:
            list_data.append(i["name"])
            counter += 1
    return list_data
```

```python
# Extract the genres and keyords name from the object array
movies_data_df["genres"] = movies_data_df["genres"].apply(extract_genres)
movies_data_df["keywords"] = movies_data_df["keywords"].apply(extract_genres)

# Extracting the year of release and storing the value in same column
movies_data_df["release_date"] = pd.to_datetime(movies_data_df["release_date"]).dt.year
```

```python
# Extracting only top 3 leading actor from the list
movies_data_df["cast"] = movies_data_df["cast"].apply(lambda x: extract_genres(x, 3))
```

```python
movies_data_df.head()
```

```python
# Extracting the crew member whose job is director in the list
def extract_crew(col):
    list_data = []
    for data in ast.literal_eval(col):
        if data["job"] == "Director":
            list_data.append(data["name"])
    return list_data
```

```python
movies_data_df["crew"] = movies_data_df["crew"].apply(extract_crew)
```

```python
# Coverting the overview string value in the list of string
movies_data_df["overview"] = movies_data_df["overview"].str.split()
movies_data_df
```

```python
# Now removing all the empty space from all the column
# We are doing this beacuse "Swift Taylor" will used as 2 different value "Swift" & "Taylor"
# So if there is any movie which has cast as "David Taylor", the system will recommand the movie from both cast
# Even if the movie is not related to each other
# It will only add in recommendation based on "Taylor" keyword matched in both movie
# So after removing the spacewe can uniquely identify "SwiftTaylor" and "DavidTaylor" 
movies_data_df["cast"] = movies_data_df["cast"].apply(lambda x: [i.replace(" ", "") for i in x])
movies_data_df["crew"] = movies_data_df["crew"].apply(lambda x: [i.replace(" ", "") for i in x])
movies_data_df["keywords"] = movies_data_df["keywords"].apply(lambda x: [i.replace(" ", "") for i in x])
movies_data_df["genres"] = movies_data_df["genres"].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies_data_df
```

```python
# Merged all the list column into one tag column on which we will apply the further manipulations
movies_data_df["tags"] = movies_data_df["genres"] + movies_data_df["keywords"] + movies_data_df["crew"] + movies_data_df["cast"] + movies_data_df["overview"]
```

```python
movies_data_df["tags"][0]
```

```python
# After merging the value in tags column, dropped all the unwanted column
movies_data_df = movies_data_df.drop(columns=["genres", "keywords", "cast", "crew", "overview"])
```

```python
movies_data_df.head()
```

```python
# Converted the list tags column in to string by joining the value with empty space
movies_data_df["tags"] = movies_data_df["tags"].apply(lambda x: " ".join(x))
```

```python
movies_data_df.head(3)
```

```python
# Converted the values of tag and title to lowercase for consistency
movies_data_df["tags"] = movies_data_df["tags"].apply(lambda x: x.lower())
movies_data_df["title"] = movies_data_df["title"].apply(lambda x: x.lower())
```

#### Steming

```python
# Importing porter steam from nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
# Applying the stemming techique
# Stemming can only be applied on list
# So we first converted the string to list, applied the stemming, append to the list back
# And before returning we joined the list with empty space again
def stemming(text):
    stemmed_list = []
    for i in text.split():
        stemmed_list.append(ps.stem(i))
    return " ".join(stemmed_list)
```

```python
# Applied steeming on tags column which has all the relevant data about genres, cast, crew, overview and keywords
movies_data_df["tags"] = movies_data_df["tags"].apply(stemming)
```

```python
# Importing the count vectorixer to select the feature and stop words in english
# Need to keep "english" in small case as our data is already converted to lowercase
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
# Now using the count vectorizer we transfored and fitted the tags data to get the vector
vectors = cv.fit_transform(movies_data_df["tags"])
```

```python
# Storing the vector in array
vectors = vectors.toarray()
```

```python
# This function provide all the 5000 name which is used as a feature
cv.get_feature_names_out(vectors)[:500]
```

```python
# Importing the cosin similarities and passed the vector to get the similar value array
from sklearn.metrics.pairwise import cosine_similarity

similarities = cosine_similarity(vectors)
```

```python
similarities
```

```python
def recommend_movies(movie_name):
    """Returns the top 5 movies similar to the given movie."""

    recommend_list = []  # Stores recommended movie titles

    # Find the movie in the dataset
    movie_data = movies_data_df[movies_data_df["title"] == movie_name.lower()]

    if not movie_data.empty:
        movie_index = movie_data.index[0]  # Get the movie's index

        recommendations = similarities[movie_index]  # Get similarity scores

        # Sort movies by similarity score and exclude the input movie
        movie_list = sorted(
            enumerate(recommendations),
            reverse=True,
            key=lambda x: x[1]
        )[1:6]  # Select top 5 similar movies

        # Add recommended movie titles to the list
        for movie in movie_list:
            recommend_list.append(movies_data_df.iloc[movie[0]].title)

        return recommend_list  # Return recommended movies

    else:
        return []  # Return empty list if movie is not found
```
