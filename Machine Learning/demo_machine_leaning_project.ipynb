#### Data Cleaning

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
 df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
df['Country'] = df['Country'].astype('category')
df['City'] = df['City'].astype('category')
df['University'] = df['University'].astype('category')
df['Program'] = df['Program'].astype('category')
df['Level'] = df['Level'].astype('category')
df['Duration_Years'] = df['Duration_Years'].astype('float32')
df['Rent_USD'] = df['Rent_USD'].astype('float32')
df['Visa_Fee_USD'] = df['Visa_Fee_USD'].astype('float32')
df['Insurance_USD'] = df['Insurance_USD'].astype('float32')
```

```python
df.info()
```

#### Exploratory Data Analysis

```python
top_country_list = df['Country'].value_counts().head(10).index.tolist()
top_country_df = df[df['Country'].isin(top_country_list)].copy()
top_country_df['Country'] = top_country_df['Country'].cat.remove_unused_categories()
```

```python
sns.countplot(data=top_country_df, x='Country')
plt.xticks(rotation=90)
plt.grid()
```

```python
top_city_list = df['City'].value_counts().head(10).index.tolist()
top_city_df = df[df['City'].isin(top_city_list)].copy()
top_city_df['City'] = top_city_df['City'].cat.remove_unused_categories()
```

```python
sns.countplot(data=top_city_df, x='City')
plt.xticks(rotation=90)
plt.grid()
```
