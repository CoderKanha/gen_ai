#### Train Test Split

```python
from sklearn.model_selection import train_test_split

x = data.drop(columns=["cognitive_load"])
y = data["cognitive_load"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

#### Feature Scaling

```python
data.columns
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours', 'commits', 'bugs_reported', 'ai_usage_hours']] = scaler.fit_transform(x_train[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours', 'commits', 'bugs_reported', 'ai_usage_hours']])
x_test[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours', 'commits', 'bugs_reported', 'ai_usage_hours']] = scaler.transform(x_test[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours', 'commits', 'bugs_reported', 'ai_usage_hours']])
```

```python
x_train
```

#### Building Modal

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Dense(64, input_shape=(x_train.shape[1],), activation="linear"),
    Dense(32, activation="linear"),
    Dense(1, activation="linear"),
])
```

```python
model.summary()
```

```python
model.compile(optimizer="adam", metrics=["mae"], loss="mse")
```

```python
stopping_callback = EarlyStopping(monitor="loss", patience=5, restore_best_weights=True)
```

```python
model.fit(x_train, y_train, epochs=100, callbacks=[stopping_callback])
```

```python
from sklearn.metrics import mean_squared_error

y_predict = model.predict(x_test)

mean_squared_error(y_test, y_predict)
```
