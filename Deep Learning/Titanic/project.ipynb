```python
import pandas as pd
```

```python
data = pd.read_csv("huge_1M_titanic.csv")
```

```python
data.drop(columns=["PassengerId", "Name", "Age", "Ticket", "Cabin"], inplace=True)
```

#### Data Cleaning

```python
data["Embarked"] = data["Embarked"].replace({"S": "Southampton", "C": "Chebourg", "Q": "Queenstown"})
```

```python
data.head()
```

```python
data.dropna(subset=["Embarked"], inplace=True)
```

```python
data.isnull().sum()
```

```python
data.shape
```

```python
data["Fare"]=data["Fare"].round(2)
```

#### Data Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data["Sex"] = le.fit_transform(data["Sex"])
```

```python
data.shape
```

```python
from sklearn.preprocessing import OneHotEncoder

oeh = OneHotEncoder(sparse_output=False)
```

```python
embarked = oeh.fit_transform(data[["Embarked"]])
```

```python
embarked = pd.DataFrame(embarked, columns=oeh.get_feature_names_out())
```

```python
embarked
```

```python
data = data.reset_index(drop=True)
data = pd.concat([data.drop(columns=["Embarked"]), embarked], axis=1)
```

```python
from sklearn.preprocessing import StandardScaler

scale = StandardScaler()
```

```python
data.columns
num_columns = ['Pclass', 'SibSp', 'Parch', 'Fare']
```

```python
data[num_columns] = scale.fit_transform(data[num_columns])
```

```python
x = data.drop(columns=["Survived"])
```

```python
y = data["Survived"]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
x_train, x_valid, y_train, y_valid = train_test_split(x_train, y_train, test_size=0.2, random_state=42)
```

```python
x_train.shape
```

```python
x_valid.shape
```

#### Model Building Step

```python
!python -m pip install tensorflow tensorboard
```

```python
import tensorflow
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Dense 
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Dense(64, input_shape=(x.shape[1],), activation="relu"),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid"),
])
```

```python
model.summary()
```

```python
opt = tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
model.compile(optimizer=opt, loss="binary_crossentropy", metrics=["accuracy"])
```

```python
stopping_callback = EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
```

```python
model.fit(x_train, y_train, validation_data=(x_valid, y_valid), epochs=10, callbacks=[stopping_callback])
```

```python
model.save("model.h5")
model.save("model.keras")
```
