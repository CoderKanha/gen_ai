#### Feature Enineering

```python
df["Total Cost"] = df["Tuition_USD"] + df["Rent_USD"]
```

```python
df.head()
```

```python
# Feature Selection --> Mention all the column names inside the list
# and that list should be inside the dataframe
```

```python
df.columns
```

```python
df = df[['Country', 'City', 'University', 'Program', 'Level', 'Duration_Years', 'Living_Cost_Index', 'Visa_Fee_USD', 'Insurance_USD', 'Total Cost']]
```

```python
df.head()
```

#### Feature Encoding

##### Convert categorial column into numerical column in order to feed the data inside the ML model

##### Type of encoder are:
##### 1. Ordinal Encode - If the value in the column can be arrange in some sort of order from high to low or vice versa

##### 2. One Hot Encoder - If the value in the column can't be arrange in the order, each category is independent with each other

##### 3. Label Encoder - It is used to convert the result or output column in numberical format

```python
# Importing the label encoder from the sk-learn lib
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
df['Country'] = le.fit_transform(df['Country'])
df['City'] = le.fit_transform(df['City'])
df['University'] = le.fit_transform(df['University'])
df['Program'] = le.fit_transform(df['Program'])
df['Level'] = le.fit_transform(df['Level'])
```

```python
df.head()
```

```python
df.describe()
```

##### Separating Input and Output Data

```python
X = df.iloc[:, :9]
Y = df["Total Cost"]
```

```python
X
```

```python
Y
```

#### Feature Scaling

##### Scale all column values in a similar range.

```python
# importing StandardScaler

from sklearn.preprocessing import StandardScaler

s = StandardScaler()
```

```python
X = s.fit_transform(X)
X
```

#### Train Test split data

##### 80% -> Training
##### 20% -> Testing

```python
# Importing train_test_split from sklearn

from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_Train, Y_Test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
Y_predict = lr.predict(X_test)
```

```python
# Evaluation Maties ->
# 1. Mean Square Error
# 2. Root Mean Square Error
# 3. Mean Absolute Error
# 4. R2 Score
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

Stacking Regressor

```python
from sklearn.ensemble import RandomForestRegressor, StackingRegressor, BaggingRegressor
from sklearn.tree import DecisionTreeRegressor

estimators = [('rf', RandomForestRegressor()), ('br', BaggingRegressor())]
```

```python
stack = StackingRegressor(estimators=estimators, final_estimator=DecisionTreeRegressor())
```

```python
stack.fit(X_train, Y_Train)
```

```python
Y_predict = stack.predict(X_test)
```

```python
print(f"mean_absolute_error: {mean_absolute_error(Y_Test, Y_predict)}")
print(f"mean_squared_error: {mean_squared_error(Y_Test, Y_predict)}")
print(f"root_mean_squared_error: {root_mean_squared_error(Y_Test, Y_predict)}")
print(f"r2_score: {r2_score(Y_Test, Y_predict)}")
```
