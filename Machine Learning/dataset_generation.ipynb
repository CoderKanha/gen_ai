```python

import pandas as pd
import numpy as np

np.random.seed(42)

n = 2500

products = {
    "Laptop": "Electronics",
    "Smartphone": "Electronics",
    "Headphones": "Electronics",
    "Office Chair": "Furniture",
    "Desk": "Furniture",
    "Water Bottle": "Home & Kitchen",
    "Coffee Maker": "Home & Kitchen",
    "Running Shoes": "Sports",
    "Yoga Mat": "Sports",
    "Backpack": "Accessories",
    "Wrist Watch": "Accessories",
    "Novel": "Books"
}

reviews = [
    "Fast delivery and good quality",
    "Product met my expectations",
    "Packaging could be improved",
    "Great value for money",
    "Customer support was helpful",
    "Delivery was slightly delayed",
    "Would recommend this product",
    "Item was exactly as described",
    "Good product for daily use",
    "Not satisfied with the finish",
    "Easy ordering experience",
    "Excellent quality and service"
]

cities = ["Anand", "Ahmedabad", "Vadodara",
          "Surat", "Rajkot", "Mumbai",
          "Pune", "Jaipur", "Delhi", "Bengaluru"]

data = []

for i in range(n):

    product = np.random.choice(list(products.keys()))
    category = products[product]

    quantity = np.random.randint(1, 8)
    price = round(np.random.uniform(10, 1500), 2)
    age = np.random.randint(18, 71)

    date = pd.Timestamp("2022-01-01") + pd.Timedelta(
        days=int(np.random.randint(0, 1460))
    )

    data.append({
        "Order_ID": f"ORD{i+1:05d}",
        "Order_Date": date,
        "Customer_Age": age,
        "Product": product,
        "Category": category,
        "Quantity": quantity,
        "Unit_Price": price,
        "Discount_Rate": np.random.choice(
            [0, 0.05, 0.10, 0.15, 0.20]
        ),
        "Unit_Cost": round(price * np.random.uniform(0.45, 0.78), 2),
        "City": np.random.choice(cities),
        "Sales_Channel": np.random.choice(
            ["Online", "In-store", "Mobile App", "Marketplace"]
        ),
        "Payment_Method": np.random.choice(
            ["Credit Card", "Debit Card", "UPI", "Cash", "Net Banking"]
        ),
        "Customer_Review": np.random.choice(reviews)
    })

df = pd.DataFrame(data)

# Save CSV
df.to_csv("retail_sales_dataset.csv", index=False)

print("Dataset created successfully!")
print("Total records:", len(df))
print(df.head())
```
