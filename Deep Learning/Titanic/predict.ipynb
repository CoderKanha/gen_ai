```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open("label_encoder.pkl", "rb") as file:
    label = pickle.load(file)
```

```python
with open("one_hot_encoder.pkl", "rb") as file:
    onehot = pickle.load(file)
```

```python
with open("scaler.pkl", "rb") as file:
    scaler = pickle.load(file)
```

```python
model = load_model("model.h5")
```

```python
user1 = {
    "Pclass": 3,
    "Sex": "female",
    "SibSp": 1,
    "Parch": 1,
    "Fare": 10.80,
    "Embarked": "Queenstown"
}

user2 = {
    "Pclass": 2,
    "Sex": "male",
    "SibSp": 3,
    "Parch": 2,
    "Fare": 20.56,
    "Embarked": "Southampton"
}

user3 = {
    "Pclass": 1,
    "Sex": "female",
    "SibSp": 0,
    "Parch": 0,
    "Fare": 90.77,
    "Embarked": "Chebourg"
}
```

```python
user = pd.DataFrame([user1, user2, user3])
```

```python
user
```

```python
user["Sex"] = label.transform(user["Sex"])
```

```python
embarked = onehot.transform(user[["Embarked"]])
```

```python
user
```

```python
embarked = pd.DataFrame(embarked, columns=onehot.get_feature_names_out())
```

```python
embarked
```

```python
user = pd.concat([user.drop(columns=["Embarked"]), embarked], axis=1)
```

```python
user
```

```python
user[["Pclass", "SibSp", "Parch", "Fare"]] = scaler.transform(user[["Pclass", "SibSp", "Parch", "Fare"]])
```

```python
user
```

```python
y = model.predict(user)
```
