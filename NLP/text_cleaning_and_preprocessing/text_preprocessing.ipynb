Split Tokenizer

```python
sample_text = "This is my new python notebook . I am learning how tokenizer work! My email id is test@gmail.com. I am leaving new office where 5km ride cost me $20. I am very new to New-Delhi and A.I. field"
```

```python
# Word tokenizer
sample_text.split()
```

```python
# Sentence tokenizer
sample_text.split(".")
```

Regular Expression Based Tokenizer

```python
import re

re.findall("[\\w']+", sample_text)
```

NLTK -> Natual Language Tool Kit

```python
# import nltk
# nltk.download("punkt_tab")
```

```python
# Word and sentence tokenizer
from nltk.tokenize import word_tokenize, sent_tokenize
word_tokenize(sample_text)
```

```python
sent_tokenize(sample_text)
```

Spacy Tokenizer

```python
import spacy
```

```python
# from spacy.cli import download
# download("en_core_web_sm")
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
spacy_token = nlp(sample_text)
```

```python
for i in spacy_token:
    print(i)
```

Stemming

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

for i in sample_text.split():
    print(ps.stem(i))
```

```python
from nltk.stem import WordNetLemmatizer

lematizer = WordNetLemmatizer()

for i in sample_text.split():
    print(lematizer.lemmatize(i, pos="v"))
```
