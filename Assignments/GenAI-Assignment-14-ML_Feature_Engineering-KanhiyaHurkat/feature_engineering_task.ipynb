#### Importing required libraries and dataset

```python
# Importing required library
import pandas as pd

# Importing required dataset
retail_sales_df = pd.read_csv("retail_sales_dataset.csv")
retail_sales_df.sample(5)
```

#### Task 1 - Creating New Features

```python
# Converting the order date column to datatime
retail_sales_df["Order_Date"] = pd.to_datetime(retail_sales_df["Order_Date"])

# Creating meaningful features
# Total value = quantity x price
retail_sales_df["Total_Value"] = retail_sales_df["Quantity"] * retail_sales_df["Unit_Price"]
 
# Revenue (after discount)
retail_sales_df["Revenue"] = retail_sales_df["Total_Value"] * (1 - retail_sales_df["Discount_Rate"])
 
# Total cost and profit
retail_sales_df["Total_Cost"] = retail_sales_df["Quantity"] * retail_sales_df["Unit_Cost"]
retail_sales_df["Profit"] = retail_sales_df["Revenue"] - retail_sales_df["Total_Cost"]
```

#### Task 2 - Handling Date and Text Features

```python
# Extracting Year, Month and Day from the order date
retail_sales_df["Order_Year"] = retail_sales_df["Order_Date"].dt.year
retail_sales_df["Order_Month"] = retail_sales_df["Order_Date"].dt.month
retail_sales_df["Order_Day"] = retail_sales_df["Order_Date"].dt.day

# Counting review string lenght and word count
retail_sales_df["Review_Lenth"] = retail_sales_df["Customer_Review"].apply(lambda x: len(x))
retail_sales_df["Review_Word_Count"] = retail_sales_df["Customer_Review"].apply(lambda x: len(x.split()))
```

```python
# Keeping only required columns
final_retail_sales_df = retail_sales_df[["Customer_Age", "Product", "Category", "Quantity", "Unit_Price", "Discount_Rate", "Unit_Cost", "City", "Sales_Channel", "Payment_Method", "Customer_Review", "Total_Value", "Revenue", "Total_Cost", "Profit", "Order_Year", "Order_Month"]]
```

```python
# Using the appropriate datatype to each column to reduce the memory usage
final_retail_sales_df["Customer_Age"] = final_retail_sales_df["Customer_Age"].astype("int32")
final_retail_sales_df["Product"] = final_retail_sales_df["Product"].astype("category")
final_retail_sales_df["Category"] = final_retail_sales_df["Category"].astype("category")
final_retail_sales_df["City"] = final_retail_sales_df["City"].astype("category")
final_retail_sales_df["Sales_Channel"] = final_retail_sales_df["Sales_Channel"].astype("category")
final_retail_sales_df["Payment_Method"] = final_retail_sales_df["Payment_Method"].astype("category")
```

#### Task 7 - Create Preprocessing Pipeline

```python
# Step 1 - Separating numerical, nominal and ordinal column

numerical_column = final_retail_sales_df.select_dtypes(include=["number"]).columns.tolist()

nominal_column = ["Product", "Category", "City", "Sales_Channel", "Payment_Method", "Customer_Review"]

ordinal_column = [] # No ordial column available in the dataset
```

```python
# Step 2 - Creating the pipeline for each type of data

# Numerical pipeline
numerical_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="mean")), 
    ("scaling", StandardScaler())
])
```

```python
# Nominal pipeline
nominal_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encoding", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
# Step 3 - Combine pipeline transformer w.r.t. data
# Column Transformer
transformer = ColumnTransformer(transformers=[
    ("Numeric Transformer", numerical_pipeline, numerical_column),
    ("Nominal Transformer", nominal_pipeline, nominal_column),
])

transformer
```

#### Task 8 - Full Scikit-learn Pipeline

```python
# Importing the required packages for creating the pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score, accuracy_score, precision_score
```

```python
# Injecting the required dataset
electronic_store_df = pd.read_csv("electronic_store_sample_data.csv")
electronic_store_df.sample()
```

```python
# Changing the datatypes, adding new features
electronic_store_df["Age"] = electronic_store_df["Age"].astype("int32")
electronic_store_df["Rating"] = electronic_store_df["Rating"].astype("int32")

electronic_store_df["Purchase Date"] = pd.to_datetime(electronic_store_df["Purchase Date"])
electronic_store_df["Purchase Year"] = electronic_store_df["Purchase Date"].dt.year
electronic_store_df["Purchase Month"] = electronic_store_df["Purchase Date"].dt.month

electronic_store_df["Gender"] = electronic_store_df["Gender"].astype("category")
electronic_store_df["Loyalty Member"] = electronic_store_df["Loyalty Member"].astype("category")
electronic_store_df["Product Type"] = electronic_store_df["Product Type"].astype("category")
electronic_store_df["Order Status"] = electronic_store_df["Order Status"].astype("category")
electronic_store_df["Payment Method"] = electronic_store_df["Payment Method"].astype("category")
electronic_store_df["Shipping Type"] = electronic_store_df["Shipping Type"].astype("category")
```

```python
# Checking the dataset structure
electronic_store_df.info()
```

Linear Regression

```python
# Separating the numerical, ordinal, nominal and target column
electronic_sales_num_col = ["Age", "Rating", "Unit Price", "Quantity", "Purchase Year", "Purchase Month"]

electronic_sales_ordinal_col = ["Order Status"]

electronic_sales_nominal_col = ["Gender", "Loyalty Member", "Product Type", "Payment Method", "Shipping Type"]

electronic_sales_target_col = ["Total Price"]
```

```python
# Creating the pipeline for each kind of column
electronic_sales_num_pipeline = Pipeline(steps=[
    ("Impute", SimpleImputer(strategy="mean")),
    ("Scaling", StandardScaler())
])

electronic_sales_ordinal_pipeline = Pipeline(steps=[
    ("Impute", SimpleImputer(strategy="most_frequent")),
    ("Encoding", OrdinalEncoder(categories=[["Cancelled", "Completed"]])),
    ("Scaling", StandardScaler())
])

electronic_sales_nominal_pipeline = Pipeline(steps=[
    ("Impute", SimpleImputer(strategy="most_frequent")),
    ("Encoding", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
    ("Scaling", StandardScaler())
])
```

```python
# Creating the column transformer

electronic_sales_transformer = ColumnTransformer(transformers=[
    ("Numeric Transformation", electronic_sales_num_pipeline, electronic_sales_num_col),
    ("Ordinal Transformation", electronic_sales_ordinal_pipeline, electronic_sales_ordinal_col),
    ("Nominal Transformation", electronic_sales_nominal_pipeline, electronic_sales_nominal_col),
])

electronic_sales_transformer
```

```python
# Creating the model
electronic_sales_model = Pipeline(steps=[
    ("Transformer", electronic_sales_transformer),
    ("Regression", LinearRegression())
])

electronic_sales_model
```

```python
# Separating the feature and target column
x = ["Age", "Gender", "Loyalty Member", "Product Type", "Rating", "Order Status", "Payment Method", "Unit Price", "Quantity", "Shipping Type", "Purchase Year", "Purchase Month"]
y = ["Total Price"]
```

```python
# Spliting the dataset into train and test part
x_train, x_test, y_train, y_test = train_test_split(electronic_store_df[x], electronic_store_df[y], test_size=0.2)
```

```python
# Training model
electronic_sales_model.fit(x_train, y_train)
```

```python
# Predicting the value using the test dataset
y_predict = electronic_sales_model.predict(x_test)
```

```python
# Checking the predection score for the regression model
print(f"mean_absolute_error: {mean_absolute_error(y_test, y_predict)}")
print(f"mean_squared_error: {mean_squared_error(y_test, y_predict)}")
print(f"root_mean_squared_error: {root_mean_squared_error(y_test, y_predict)}")
print(f"r2_score: {r2_score(y_test, y_predict)}")
```

Logistic Regression 

```python
# Separating the numerical, ordinal, nominal and target column
classification_num_col = ["Age", "Unit Price", "Quantity", "Purchase Year", "Purchase Month", "Total Price"]

classification_ordinal_col = ["Order Status"]

classification_nominal_col = ["Gender", "Loyalty Member", "Product Type", "Payment Method", "Shipping Type"]

classification_target_col = ["Rating"]
```

```python
classification_num_pipeline = Pipeline(steps=[
    ("Impute", SimpleImputer(strategy="mean")),
    ("Scaling", StandardScaler())
])

classification_ordinal_pipeline = Pipeline(steps=[
    ("Impute", SimpleImputer(strategy="most_frequent")),
    ("Encoding", OrdinalEncoder(categories=[["Cancelled", "Completed"]])),
    ("Scaling", StandardScaler())
])

classification_nominal_pipeline = Pipeline(steps=[
    ("Impute", SimpleImputer(strategy="most_frequent")),
    ("Encoding", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
    ("Scaling", StandardScaler())
])
```

```python
# Creating the column transformer

classification_transformer = ColumnTransformer(transformers=[
    ("Numeric Transformation", classification_num_pipeline, classification_num_col),
    ("Ordinal Transformation", classification_ordinal_pipeline, classification_ordinal_col),
    ("Nominal Transformation", classification_nominal_pipeline, classification_nominal_col),
])

classification_transformer
```

```python
# Creating the model
classification_model = Pipeline(steps=[
    ("Transformer", classification_transformer),
    ("Classification", LogisticRegression())
])

classification_model
```

```python
# Separating the feature and target column
classification_x = ["Age", "Gender", "Loyalty Member", "Product Type", "Order Status", "Payment Method", "Unit Price", "Quantity", "Shipping Type", "Purchase Year", "Purchase Month", "Total Price"]
classification_y = ["Rating"]
```

```python
# Spliting the dataset into train and test part
classification_x_train, classification_x_test, classification_y_train, classification_y_test = train_test_split(electronic_store_df[classification_x], electronic_store_df[classification_y], test_size=0.2)
```

```python
# Training model
classification_model.fit(classification_x_train, classification_y_train)
```

```python
# Predicting the value using the test dataset
classification_y_predict = classification_model.predict(classification_x_test)
```

```python
# Checking the accuracy score for the classification problem
accuracy_score(classification_y_test, classification_y_predict)
```

#### Task 9 - Pipeline Benefits

1. Why are pipelines important in ML?

A pipeline puts all the steps (encoding, scaling, model training) into one chain that runs in the same order every time. This keeps the code clean and easy to repeat. It also means the same steps are applied to both training and new data, so we don’t have to remember what we did earlier.

2. What problems do pipelines solve?

The main one is data leakage. If we scale the whole dataset before splitting it, the test data influences the training, and the results look better than they really are. A pipeline learns from the training data only and then applies that to the test data. It also stops us from forgetting a step or doing the steps in the wrong order. And it makes it much easier to put a model into use later, because the preprocessing and the model are saved together.

3. Manual preprocessing vs pipeline-based preprocessing

Manual preprocessing:
-> Each step is written and run separately <br>
-> Easy to forget or repeat a step on new data <br>
-> Higher risk of data leakage <br>
-> Longer code, harder to reuse <br>
-> Hard to tune with cross-validation <br>

Pipeline-based preprocessing:
-> All steps are combined in one object <br>
-> Same steps run automatically every time <br>
-> Fit on training data only, so leakage is avoided <br>
-> Short code, easy to reuse and share <br>
-> Works directly with cross_val_score and GridSearchCV <br>

In short: manual preprocessing works for a quick test, but pipelines are the safer and cleaner way once the project gets bigger.
