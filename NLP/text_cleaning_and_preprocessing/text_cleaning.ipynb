```python
# Importing the required libraries

import pandas as pd
import re
```

```python
train_data_cleaning_df = pd.read_csv("train_data_cleaning.csv")
```

#### Lowering the case of the text

```python
train_data_cleaning_df["text_lowercase"] = train_data_cleaning_df["text"].apply(lambda x: x.lower())
```

#### Removing the dirty data

Removing the HTML tags

```python
def remove_html_tags(text):
    return re.sub(r"<.*?>", "", text)
```

Removing the URL's

```python
def remove_url(text):
    return re.sub(r"https?://\S+|www\.\S+","",text)
```

Removing Puntuation Marks

```python
def remove_puntuals(text):
    return re.sub(r"[^\w\s]", "", text)
```

```python
train_data_cleaning_df["text_lowercase"] = train_data_cleaning_df["text_lowercase"].apply(remove_puntuals)
```

```python
slang_dict = {"gn": "Good Night", "gm": "Good Morning", "yolo": "You Only Live Once", "asap": "As Soon As Possible"}

slang_data = pd.DataFrame({"slang": ["Hey ! gm", "Chill yaar, yolo", "please do it asap"]})
slang_data
```

```python
def expand_slang(text, slang_dict):
    words = text.split()
    return " ".join([slang_dict.get(word, word) for word in words])
```
