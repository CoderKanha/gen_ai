#### Task 6 Data Cleaning

```python
# Creating a copy of actual data to add randomly missing values

# Copying the oriiginal dataframe
df_missing = df.copy()

# Set random seed for reproducibility
np.random.seed(42)

# Percentage of values to remove
missing_percentage = 0.10

# Generate a random mask
random_mask = np.random.rand(*df_missing.shape) < missing_percentage

# Replace randomly selected values with NaN
df_missing = df_missing.mask(random_mask)

# Display updated missing values
df_missing.isna().sum()
```

```python
# Filling the empty cells with mean, median, mode or "Unknown"

df_missing.fillna({ "Country": df_missing["Country"].mode()[0]}, inplace=True)
df_missing.fillna({ "City": df_missing["City"].mode()[0]}, inplace=True)
df_missing.fillna({ "University": df_missing["University"].mode()[0]}, inplace=True)
df_missing.fillna({ "Program": "Unknown"}, inplace=True)
df_missing.fillna({ "Level": "Unknown"}, inplace=True)
df_missing.fillna({ "Duration_Years": df_missing["Duration_Years"].mode()[0]}, inplace=True)
df_missing.fillna({ "Tuition_USD": df_missing["Tuition_USD"].mode()[0]}, inplace=True)
df_missing.fillna({ "Living_Cost_Index": df_missing["Living_Cost_Index"].median()}, inplace=True)
df_missing.fillna({ "Rent_USD": df_missing["Rent_USD"].median()}, inplace=True)
df_missing.fillna({ "Visa_Fee_USD": df_missing["Visa_Fee_USD"].median()}, inplace=True)
df_missing.fillna({ "Insurance_USD": df_missing["Insurance_USD"].median()}, inplace=True)
df_missing.fillna({ "Exchange_Rate": df_missing["Exchange_Rate"].median()}, inplace=True)

df_missing.isna().sum()
```

```python
# Checking if there is any duplicate values in the dataframe
# If any then removing the duplicate values

df.duplicated().sum()
df.drop_duplicates(keep=False, inplace=True)
```

```python
# Renaming the columns to lowercase and snake_case

df.rename(columns={
    "Country": "country",
    "City": "city",
    "University": "university",
    "Program": "program",
    "Level": "level",
    "Duration_Years": "duration_years",
    "Tuition_USD": "tuition_usd",
    "Living_Cost_Index": "living_cost_index",
    "Rent_USD": "rent_usd",
    "Visa_Fee_USD": "visa_fee_usd",
    "Insurance_USD": "insurance_usd",
    "Exchange_Rate": "exchange_rate",
}, inplace=True)
```

```python
# Changing data types of the column
# Reduced memory consumption from 135.6 KB to 60 KB

df["country"] = df["country"].astype("category")
df["city"] = df["city"].astype("category")
df["university"] = df["university"].astype("category")
df["program"] = df["program"].astype("category")
df["level"] = df["level"].astype("category")
df["duration_years"] = df["duration_years"].astype("float32")
df["tuition_usd"] = df["tuition_usd"].astype("int32")
df["living_cost_index"] = df["living_cost_index"].astype("float32")
df["rent_usd"] = df["rent_usd"].astype("int32")
df["visa_fee_usd"] = df["visa_fee_usd"].astype("int32")
df["insurance_usd"] = df["insurance_usd"].astype("int32")
df["exchange_rate"] = df["exchange_rate"].astype("float32")
```

```python
# Splitting data based on usage of each gender on different platforms
primary_platform_snapchat = social_media_df[social_media_df["Primary_Platform"] == "Snapchat"]
primary_platform_instagram = social_media_df[social_media_df["Primary_Platform"] == "Instagram"]
primary_platform_tiktok = social_media_df[social_media_df["Primary_Platform"] == "TikTok"]
primary_platform_youtube = social_media_df[social_media_df["Primary_Platform"] == "YouTube"]
primary_platform_linkedIn = social_media_df[social_media_df["Primary_Platform"] == "LinkedIn"]
primary_platform_reddit = social_media_df[social_media_df["Primary_Platform"] == "Reddit"]
primary_platform_twitter = social_media_df[social_media_df["Primary_Platform"] == "X (Twitter)"]

gender_values = ["Female", "Male", "Non-Binary", "Prefer not to say"]

primary_platform_snapchat_df = primary_platform_snapchat.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
primary_platform_instagram_df = primary_platform_instagram.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
primary_platform_tiktok_df = primary_platform_tiktok.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
primary_platform_youtube_df = primary_platform_youtube.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
primary_platform_linkedIn_df = primary_platform_linkedIn.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
primary_platform_reddit_df = primary_platform_reddit.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
primary_platform_twitter_df = primary_platform_twitter.groupby("Gender")["Daily_Usage_Hours"].sum().reindex(gender_values, fill_value=0)
```

```python
# Ploting the Multi Bar Chart
plt.figure(figsize=(10,5))
plt.title('Daily Usage on Each Platform Based on Gender')
plt.xlabel('Gender')
plt.ylabel('Daily Usage (Hours)')
plt.bar(np.arange(primary_platform_snapchat_df.shape[0]) - 0.3, primary_platform_snapchat_df.values, label="Snapchat", width=0.1)
plt.bar(np.arange(primary_platform_instagram_df.shape[0]) - 0.2, primary_platform_instagram_df.values, label="Instagram", width=0.1)
plt.bar(np.arange(primary_platform_tiktok_df.shape[0]) - 0.1, primary_platform_tiktok_df.values, label="Tiktok", width=0.1)
plt.bar(np.arange(primary_platform_youtube_df.shape[0]), primary_platform_youtube_df.values, label="YouTube", width=0.1)
plt.bar(np.arange(primary_platform_linkedIn_df.shape[0]) + 0.1, primary_platform_linkedIn_df.values, label="LinkedIn", width=0.1)
plt.bar(np.arange(primary_platform_reddit_df.shape[0]) + 0.2, primary_platform_reddit_df.values, label="Reddit", width=0.1)
plt.bar(np.arange(primary_platform_twitter_df.shape[0]) + 0.3, primary_platform_twitter_df.values, label="X (Twitter)", width=0.1)
plt.xticks(ticks=[0, 1, 2, 3], labels=['Female', 'Male', 'Non-Binary', 'Prefer not to say'])
plt.legend()
plt.grid()
plt.show()
```

```python
# Ploting the Stacked Bar Chart
plt.figure(figsize=(10,5))
plt.title('Daily Usage on Each Platform Based on Gender')
plt.xlabel('Gender')
plt.ylabel('Daily Usage (Hours)')
plt.bar(primary_platform_snapchat_df.index, primary_platform_snapchat_df.values, label="Snapchat", width=0.5)
plt.bar(primary_platform_instagram_df.index, primary_platform_instagram_df.values, label="Instagram", bottom=primary_platform_snapchat_df.values, width=0.5)
plt.bar(primary_platform_tiktok_df.index, primary_platform_tiktok_df.values, label="Tiktok", bottom=(primary_platform_snapchat_df.values + primary_platform_instagram_df.values), width=0.5)
plt.bar(primary_platform_youtube_df.index, primary_platform_youtube_df.values, label="YouTube", bottom=(primary_platform_snapchat_df.values + primary_platform_instagram_df.values + primary_platform_tiktok_df.values), width=0.5)
plt.bar(primary_platform_linkedIn_df.index, primary_platform_linkedIn_df.values, label="LinkedIn", bottom=(primary_platform_snapchat_df.values + primary_platform_instagram_df.values + primary_platform_tiktok_df.values + primary_platform_youtube_df.values), width=0.5)
plt.bar(primary_platform_reddit_df.index, primary_platform_reddit_df.values, label="Reddit", bottom=(primary_platform_snapchat_df.values + primary_platform_instagram_df.values + primary_platform_tiktok_df.values + primary_platform_youtube_df.values + primary_platform_linkedIn_df.values), width=0.5)
plt.bar(primary_platform_twitter_df.index, primary_platform_twitter_df.values, label="X (Twitter)", bottom=(primary_platform_snapchat_df.values + primary_platform_instagram_df.values + primary_platform_tiktok_df.values + primary_platform_youtube_df.values + primary_platform_linkedIn_df.values + primary_platform_reddit_df.values), width=0.5)
plt.xticks(ticks=[0, 1, 2, 3], labels=['Female', 'Male', 'Non-Binary', 'Prefer not to say'])
plt.legend()
plt.grid()
plt.show()
```
